```python
# ============================================================
# HOUSE PRICE PREDICTION
# Machine Learning Regression Project
# Google Colab
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

housing = fetch_california_housing(as_frame=True)

df = housing.frame.copy()

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 3. DATA EXPLORATION
# ============================================================

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe())


# ============================================================
# 4. DATA CLEANING
# ============================================================

df = df.drop_duplicates()

df = df.dropna()

print("\nDataset after cleaning:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# 5. TARGET AND FEATURES
# ============================================================

X = df.drop(columns=["MedHouseVal"])
y = df["MedHouseVal"]

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print("MedHouseVal")


# ============================================================
# 6. EXPLORATORY DATA ANALYSIS
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    y,
    bins=40,
    kde=True
)

plt.title("Distribution of House Prices")
plt.xlabel("Median House Value")
plt.ylabel("Frequency")
plt.show()


# ============================================================
# 7. CORRELATION ANALYSIS
# ============================================================

plt.figure(figsize=(10, 7))

correlation_matrix = df.corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()


# ============================================================
# 8. FEATURE VS HOUSE PRICE
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="MedInc",
    y="MedHouseVal",
    alpha=0.3
)

plt.title("Median Income vs House Value")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.show()


# ============================================================
# 9. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


# ============================================================
# 10. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ============================================================
# 11. LINEAR REGRESSION MODEL
# ============================================================

linear_model = LinearRegression()

linear_model.fit(
    X_train_scaled,
    y_train
)

linear_predictions = linear_model.predict(
    X_test_scaled
)

print("\nLinear Regression model trained successfully!")


# ============================================================
# 12. LINEAR REGRESSION EVALUATION
# ============================================================

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)

print("\n============================================================")
print("LINEAR REGRESSION PERFORMANCE")
print("============================================================")

print("MAE :", round(linear_mae, 4))
print("RMSE:", round(linear_rmse, 4))
print("R²  :", round(linear_r2, 4))


# ============================================================
# 13. RANDOM FOREST REGRESSION
# ============================================================

random_forest = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

random_forest.fit(
    X_train,
    y_train
)

rf_predictions = random_forest.predict(
    X_test
)

print("\nRandom Forest model trained successfully!")


# ============================================================
# 14. RANDOM FOREST EVALUATION
# ============================================================

rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

print("\n============================================================")
print("RANDOM FOREST PERFORMANCE")
print("============================================================")

print("MAE :", round(rf_mae, 4))
print("RMSE:", round(rf_rmse, 4))
print("R²  :", round(rf_r2, 4))


# ============================================================
# 15. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest"
    ],
    "MAE": [
        linear_mae,
        rf_mae
    ],
    "RMSE": [
        linear_rmse,
        rf_rmse
    ],
    "R2": [
        linear_r2,
        rf_r2
    ]
})

print("\n============================================================")
print("MODEL COMPARISON")
print("============================================================")

display(
    comparison.round(4)
)


# ============================================================
# 16. SELECT BEST MODEL
# ============================================================

if rf_r2 > linear_r2:
    best_model_name = "Random Forest"
    best_predictions = rf_predictions
else:
    best_model_name = "Linear Regression"
    best_predictions = linear_predictions

print(
    "\nBest model based on R²:",
    best_model_name
)


# ============================================================
# 17. ACTUAL VS PREDICTED VALUES
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.4
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.title(
    f"Actual vs Predicted House Values - {best_model_name}"
)

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")

plt.tight_layout()
plt.show()


# ============================================================
# 18. RESIDUAL ANALYSIS
# ============================================================

residuals = y_test - best_predictions

plt.figure(figsize=(8, 5))

sns.histplot(
    residuals,
    bins=40,
    kde=True
)

plt.title(
    f"Prediction Residuals - {best_model_name}"
)

plt.xlabel("Residual")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 19. RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": random_forest.feature_importances_
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

print("\nFeature importance:")
display(feature_importance)


# ============================================================
# 20. FEATURE IMPORTANCE VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance - Random Forest")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()


# ============================================================
# 21. CREATE PREDICTION DATAFRAME
# ============================================================

predictions_df = X_test.copy()

predictions_df["Actual_Price"] = y_test.values

predictions_df["Predicted_Price"] = best_predictions

predictions_df["Prediction_Error"] = (
    predictions_df["Actual_Price"]
    - predictions_df["Predicted_Price"]
)

print("\nSample predictions:")

display(
    predictions_df.head(10)
)


# ============================================================
# 22. SAVE RESULTS
# ============================================================

comparison.to_csv(
    "model_comparison.csv",
    index=False
)

feature_importance.to_csv(
    "feature_importance.csv",
    index=False
)

predictions_df.to_csv(
    "house_price_predictions.csv",
    index=False
)

print("\nOutput files saved successfully!")


# ============================================================
# 23. FINAL SUMMARY
# ============================================================

print("""
============================================================
HOUSE PRICE PREDICTION COMPLETED
============================================================

✓ Dataset loaded
✓ Data explored
✓ Data cleaned
✓ Features separated
✓ Exploratory Data Analysis completed
✓ Correlation analysis completed
✓ Train/Test split completed
✓ Feature scaling completed
✓ Linear Regression trained
✓ Random Forest trained
✓ Models evaluated
✓ Models compared
✓ Best model selected
✓ Predictions generated
✓ Feature importance analyzed
✓ Results exported

============================================================
PROJECT INFORMATION
============================================================

Project Type : Machine Learning Regression
Dataset      : California Housing
Models       : Linear Regression + Random Forest
Environment  : Google Colab
Goal         : Predict Median House Values

============================================================
""")
```
